### Group ảnh theo cover và gallery

```python
import pandas as pd

# Đọc file
df = pd.read_csv("df_new.csv")

# 1) Lấy cover: mỗi poi_id có 1 cover -> lấy url tương ứng
cover = (
    df[df["image_type"] == "cover_image_url"]
    .groupby("poi_id")["cdn_url"]
    .first()
    .rename("cover_image_url")
)

# 2) Lấy gallery: gộp tất cả url của cùng poi_id, cách nhau bởi dấu phẩy
gallery = (
    df[df["image_type"] == "gallery_urls"]
    .groupby("poi_id")["cdn_url"]
    .apply(lambda urls: ",".join(urls))
    .rename("gallery_image_url")
)

# 3) Gộp hai cột lại theo poi_id
result = pd.concat([cover, gallery], axis=1).reset_index()

# 4) Ghi ra file mới
result.to_csv("poi_images.csv", index=False)
```

### Merge hai file vào nhau để visualize

```python
import pandas as pd

# File đã gộp gallery từ lần trước
poi = pd.read_csv("poi_images.csv")

# File Food đã gắn hậu tố _M/_O
food = pd.read_csv("Food_v1_3_suffixed.csv")

# 5 cột muốn thêm (kèm cột id để ghép)
cols = ["poi_id_M", "category_l1_M", "name_M", "category_l2_M",
        "cover_image_url_M", "gallery_urls_O"]

# Ghép theo poi_id (left join -> giữ nguyên toàn bộ dòng của poi_images)
result = poi.merge(
    food[cols],
    left_on="poi_id",
    right_on="poi_id_M",
    how="left",
)

# Bỏ cột poi_id_M bị trùng sau khi ghép
result = result.drop(columns=["poi_id_M"])

result.to_csv("poi_images_merged.csv", index=False)
```
